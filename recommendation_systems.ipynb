```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
from google.colab import drive
drive.mount('/content/drive')
dataset=pd.read_csv('/content/drive/MyDrive/datasets/updated_steamgames.csv')
dataset
```

```python
features = dataset[['name', 'genre', 'game_description', 'popular_tags', 'game_details']]
features = features.fillna('')
features_list = features['name'] + ' ' + features['genre'] + ' ' + features['game_description'] + ' ' + features['popular_tags'] + ' ' + features['game_details']
```

```python
dataset = dataset.iloc[:5000]
```

```python
vectorizer = TfidfVectorizer()
vector = vectorizer.fit_transform(features_list)
```

```python
cosinesimilarity=cosine_similarity(vector)
```

```python
def recommend_games(game_name, n = 10):
  index = dataset[dataset['name'] == game_name].index[0]
  similarity_score = list(enumerate(cosinesimilarity[index]))
  similarity_score=sorted(similarity_score,key=lambda x:x[1],reverse=True)

  recommendations = []

  for i, score in similarity_score[1:n+1]:
    recommendations.append(dataset.iloc[i]['name'])

  return recommendations
```

```python
print(recommend_games("Half-Life 2"))
```

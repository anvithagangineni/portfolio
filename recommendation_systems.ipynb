```python
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')
dataset=pd.read_csv('/content/drive/MyDrive/datasets/updated_steamgames.csv')
dataset
```

```python
dataset=dataset.drop(['url'],axis=1)
dataset
```

```python
features_list=dataset['name']+dataset['genre']+dataset['game_description']+dataset['popular_tags']+dataset['game_details']
features_list=features_list.fillna('')
```

```python
for i in range(5000,10000):
  dataset.drop(index=i,inplace=True)
```

```python
path='/content/drive/MyDrive/datasets/updated_steamgames.csv'
dataset.to_csv(path)
```

```python
dataset
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer=TfidfVectorizer()
```

```python
vector=vectorizer.fit_transform(features_list)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
cosinesimilarity=cosine_similarity(vector)
```

```python
names=dataset['name'].values.tolist()
games=input('Enter game name:')
for i in names:
  if i==games:
    s=names.index(i)
print(s)
```

```python
similarity=list(enumerate(cosinesimilarity[s]))
print(similarity)
```

```python
similarity_score=sorted(similarity,key=lambda x:x[1],reverse=True)
print(similarity_score)
```

```python
count=1
for i in similarity_score:
  num=i[0]
  GAMES=dataset[dataset.index==num]['name'].values[0]
  if count<=10:
    print(count,GAMES)
    count=count+1
```
